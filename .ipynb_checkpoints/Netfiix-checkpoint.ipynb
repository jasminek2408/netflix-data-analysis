# NETFLIX DATA ANALYSIS USING NUMPY,PANDAS, AND MATPLOTLIB 

## importing libraries

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

## loading the dataset

```python
df = pd.read_csv("netflix_titles.csv")
df.head(10)
```

```python
df.info()
```

## Checking whether the dataset have missing values or not 

```python
# it returns the bool value if data is missing i.e., True or False
df.isnull()
```

```python
# it counts the number of missing values in each column
df.isnull().sum()
```

#### We can observe from the above data that <b><i>director</b></i> column has the most number of missing values

```python
df['country'].isnull().sum()
```

### Filling the missing values in "director" column with <b>unknown</b>

```python
# it means that Clean the column and overwrite the original with the cleaned version
df['country'] = df['country'].fillna('Unknown')
```

```python
df['country'].isnull().sum()
```

### inplace = True ; mean that we change the original data directly instead of creating a new copy.

### Checking whether the date_added have null or missing values

```python
df["date_added"].isnull()
```

```python
df["date_added"].isnull().sum()
```

```python
df = df.dropna(subset=['date_added'])
```

```python
df['date_added'].isnull().sum()
```

### Convert date_added into proper date format

```python
df.head(3)
```

```python
df['date_added'] = df['date_added'].str.strip()
```

```python
df.head(3)
```

```python
df['date_added'] = pd.to_datetime(df['date_added'])
```

```python
df.loc[:3,"date_added"]
```

```python
df.iloc[0:3,5:7]
```

### Creating a new column for extracting year from date_added

```python
df['year_added'] = df['date_added'].dt.year
```

```python
df[["date_added","year_added"]]
```

```python
df.head(3)
```

```python
df.loc[0:10,["date_added","year_added"]]
```

## Task 1 - Analyzing Growth over time

### Counting content per year

```python
content_trend = df['year_added'].value_counts()
```

```python
print(content_trend)
```

### Sorting the years

```python
content_trend = content_trend.sort_index()
```

```python
print(content_trend)
```

### Plotting the graph

```python
content_trend.plot(kind='line')
plt.title("Netflix Content Growth Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
content_trend.plot(kind='line', marker='o')
```

## Task 2: Genre Analysis

```python
df["listed_in"].head(4)
```

### Split genres

```python
genres = df['listed_in'].str.split(', ')
```

```python
genres.head(4)
```

### Flatten the list

```python
all_genres = [genre for sublist in genres for genre in sublist]
```

### Count genres

```python
genre_counts = pd.Series(all_genres).value_counts().head(10)
```

```python
print(genre_counts)
```

```python
genre_counts.plot(kind='bar')
plt.title("Top 10 Genres on Netflix")
plt.xticks(rotation=45)
plt.show()
```

## Task 3: Country analysis

```python
df.head(3)
```

```python
df["country"].head(10)
```

### Splitting the countries

```python
countries = df["country"].str.split(',')
```

```python
countries.head(10)
```

### Flatten the list

```python
all_countries = [country for sublist in countries for country in sublist]
```

### Count countries

```python
country_counts = pd.Series(all_countries).value_counts().head(10)
```

```python
print(country_counts)
```

### Plotting the graph

```python
country_counts.plot(kind='bar')
plt.title("Top 10 Content Producing Countries")
plt.xticks(rotation=45)
plt.show()
```

## Task- 4 : Movies vs TV Shows Analysis

```python
type_counts = df['type'].value_counts()
```

```python
print(type_counts)
```

```python
type_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title("Movies vs TV Shows on Netflix")
plt.ylabel('')
plt.show()
```

## Task 1 : Description from the output

The analysis shows that Netflix experienced minimal content growth before 2015. However, starting from 2016, there was a significant increase in the number of titles added each year, indicating a period of rapid expansion.

The platform reached its peak content addition around 2019, suggesting an aggressive content acquisition and production strategy during this period.

After 2019, there is a slight decline in the number of titles added, which may indicate a strategic shift toward quality over quantity or external factors affecting content production.

Overall, the trend highlights Netflix’s transition from a limited library to a rapidly expanding global streaming platform.

## Task 2 : Description from the output

The analysis of genre distribution shows that **International Movies** are the most dominant category on Netflix, indicating a strong focus on global content and diverse audiences.

**Dramas** and **Comedies** also form a significant portion of the catalog, suggesting that Netflix prioritizes content that appeals to a wide range of viewers and has high engagement value.

Additionally, genres such as **Documentaries**, **Action & Adventure**, and **TV Dramas** highlight the platform’s effort to maintain variety and cater to different audience preferences.

Overall, the genre distribution reflects Netflix’s strategy of combining globally diverse content with universally popular genres to maximize audience reach and engagement.


## Task 3: Country Analysis Insights

The country-wise analysis reveals that the United States contributes the highest number of titles on Netflix, highlighting its dominance as the primary content production hub.

However, countries like India and the United Kingdom also contribute significantly, indicating Netflix’s expansion into international markets and its focus on regional content production.

The presence of multiple countries in the top contributors reflects Netflix’s global strategy of diversifying its content library to cater to audiences across different regions.

Additionally, the appearance of "Unknown" values indicates missing data, which is an important data quality consideration in real-world analysis.

Overall, the analysis suggests that while Netflix is US-centric, it is rapidly evolving into a globally diversified content platform.

## Task 4 : Movies vs TV Shows Insights

The analysis shows that Movies make up a larger proportion of Netflix’s content compared to TV Shows.

This suggests that Netflix focuses on movies for quick content consumption and to attract a broad audience base.

However, TV Shows, although fewer in number, play a crucial role in maintaining long-term user engagement, as they encourage continued subscriptions through episodic content.

This balance between movies and TV shows highlights Netflix’s strategy of combining high-volume content with engagement-driven formats to maximize user retention.

Overall, the platform leverages both content types to optimize audience reach and viewing time.
